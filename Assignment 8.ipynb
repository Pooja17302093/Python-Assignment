```python
name_list = ["Pooja", "Rahul", "Anjali", "Priya", "Arjun"]

print("Name List:", name_list)
```

```python
name_list.append("Yazhini")

print(name_list)
```

```python
age_list.insert(2, 30)
print("Element",age_list)
```

```python
name_list.remove("Yazhini")

print(name_list)
```

```python
age_list.pop()

print("Last element pop", age_list)
```

```python
age_list.extend([29, 30, 26])

print("After Extend:", age_list)
```

```python
age_list.sort(reverse=True)

print("Descending Order:", age_list)
```

```python
max_age = max(age_list)
min_age = min(age_list)
total_age = sum(age_list)

print("Maximum Age:", max_age)
print("Minimum Age:", min_age)
print("Sum of All Ages:", total_age)
```
