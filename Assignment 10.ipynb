```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

```python
print("First Three Days:")
print(temperatures_w1[:3])

print("Weekend Temperatures:")
print(temperatures_w1[-2:])

print("Middle Three Days:")
print(temperatures_w1[2:5])
```

```python
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])

print("Temperatures for Two Weeks:")
print(temperatures)
```

```python
import pandas as pd
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"])
print(marks)
```

```python
print("1st Rank Mark:")
print(marks.iloc[0])
print("Top 3 Ranks:")
print(marks.loc[["Rank1", "Rank2", "Rank3"]])
print("3rd Rank Mark:")
print(marks.iloc[2])
print("Marks Greater Than 90:")
print(marks[marks > 90])
```

```python
marks.loc["Rank1"] = 100
print("Updated Marks:")
print(marks)
marks = marks.drop("Rank5")
print("After Removing Rank5:")
print(marks)
cgpa = marks / 10
print("CGPA:")
print(cgpa)
```

```python

import pandas as pd

transactions = pd.DataFrame({"TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],"ProductCategory": ["Electronics", "Clothing", "Electronics", "Furniture","Clothing", "Electronics", "Furniture", "Clothing","Furniture", "Electronics"],"Region": ["North", "South", "North", "East", "West","North", "East", "West", "South", "North"],"Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})

print(transactions)
```

```python
print("Full DataFrame:")
display(transactions)
print("First Five Rows:")
display(transactions.head())
print("Last Five Rows:")
display(transactions.tail())
print("Shape:")
print(transactions.shape)
print("Column Names:")
print(transactions.columns)
print("DataFrame Information:")
transactions.info()
```

```python
filtered_data = transactions[(transactions["Region"] == "North") &(transactions["Amount"] > 200)]
print("North Region Transactions with Amount Greater Than 200:")
display(filtered_data)
```

```python
print("Product Category Counts:")
print(transactions["ProductCategory"].value_counts())
print("Unique Regions:")
print(transactions["Region"].unique())
```

```python
region_mean = transactions.groupby("Region")["Amount"].mean()
print("Mean Amount by Region:")
print(region_mean)
```

```python
transactions = transactions[transactions["TransactionID"] != 109]
print("After Removing Transaction 109:")
display(transactions)
```

```python
transactions["Discount"] = transactions["Amount"] * 0.10
```

```python

transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165
print("After Updating Transaction 102:")
display(transactions)
```

```python
transactions = transactions.drop(columns=["Discount"])
print("Final DataFrame:")
display(transactions)
```
